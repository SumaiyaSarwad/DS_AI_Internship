```python
#imports
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, LSTM
import numpy as np
```

```python
#Model 1 (Multiclass Classification)
model = Sequential()

model.add(Dense(32, activation='relu', input_shape=(784,)))
model.add(Dense(10, activation='softmax'))

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Dummy data
X = np.random.random((1000, 784))
y = np.random.randint(10, size=(1000,))
y = keras.utils.to_categorical(y, 10)

model.fit(X, y, epochs=5, batch_size=32)
```

```python
#Model 2 (Binary Classification)
model = Sequential()

model.add(Dense(64, activation='relu', input_shape=(20,)))
model.add(Dropout(0.5))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

X = np.random.random((1000, 20))
y = np.random.randint(2, size=(1000, 1))

model.fit(X, y, epochs=5, batch_size=32)
```

```python
#Model 3 (LSTM )
model = Sequential()

model.add(LSTM(32, input_shape=(10, 64)))
model.add(Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

X = np.random.random((1000, 10, 64))
y = np.random.randint(2, size=(1000, 1))

model.fit(X, y, epochs=5, batch_size=32)
```

```python
#Functional API 
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Concatenate

input1 = Input(shape=(784,))
input2 = Input(shape=(784,))

x1 = Dense(32, activation='relu')(input1)
x2 = Dense(32, activation='relu')(input2)

merged = Concatenate()([x1, x2])
output = Dense(10, activation='softmax')(merged)

model = Model(inputs=[input1, input2], outputs=output)

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Dummy data for 2 inputs
X1 = np.random.random((1000, 784))
X2 = np.random.random((1000, 784))

y = np.random.randint(10, size=(1000,))
y = keras.utils.to_categorical(y, 10)

# Train the model
model.fit([X1, X2], y, epochs=5, batch_size=32)
```
