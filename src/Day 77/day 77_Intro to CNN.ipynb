# Convolutional Neural Networks (CNN)

CNNs are deep learning models designed for processing structured grid data such as images.  
They automatically learn spatial hierarchies of features like edges, textures, and objects.

Applications:
- Image Classification
- Object Detection
- Image Segmentation

## Why CNNs are Effective

### 1. Local Feature Detection
CNNs detect small patterns like edges, textures, and shapes.

### 2. Parameter Sharing
Same filter is applied across the image → reduces parameters.

### 3. Translation Invariance
Objects can be detected regardless of position.

```python
#Simple CNN Implementation 
import torch
import torch.nn as nn
import torch.optim as optim

# CNN Model
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(16 * 16 * 16, 10)

    def forward(self, x):
        x = self.pool(torch.relu(self.conv1(x)))
        x = x.view(-1, 16 * 16 * 16)
        x = self.fc1(x)
        return x

# Initialize model
model = SimpleCNN()
print(model)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)
```

```python
# Dummy input (batch_size=1, channels=3, 32x32 image)
x = torch.randn(1, 3, 32, 32)

output = model(x)
print("Output shape:", output.shape)
```

- CNNs automatically learn features
- Convolution extracts spatial patterns
- Pooling reduces dimensions
- Fully connected layers perform classification
- Used in real-world applications like:
  - Face recognition
  - Self-driving cars
  - Medical imaging
