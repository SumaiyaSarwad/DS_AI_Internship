```python
#imports
import torch
import torch.nn as nn
```

```python
class SimpleRNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super().__init__()
        
        self.rnn = nn.RNN(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)
    
    def forward(self, x):
        out, _ = self.rnn(x)   # RNN output
        out = self.fc(out)     # Fully connected layer
        return out
```

```python
# batch_size = 2, sequence_length = 5, input_size = 10
X = torch.randn(2, 5, 10)
```

```python
model = SimpleRNN(input_size=10, hidden_size=8, output_size=10)
```

```python
output = model(X)

print("Output shape:", output.shape)
```

```python
criterion = nn.CrossEntropyLoss()

# reshape predictions
y_pred = output.reshape(-1, 10)

# fake labels
y_true = torch.randint(0, 10, (2*5,))

loss = criterion(y_pred, y_true)

print("Loss:", loss.item())
```

✅ 1. If an RNN is used to predict the next character in a sequence, what should be the dimensionality of the output layer?
Answer:

The output layer dimensionality should be equal to the size of the vocabulary.

In a character-level model → number of unique characters

In a word-level model → number of unique words

This is because the model predicts a probability distribution over all possible next tokens.

👉 Example:

If vocabulary = 26 letters → output size = 26

✅ 2. Why are RNNs capable of modeling dependencies across long sequences?
Answer:

RNNs use a hidden state (memory) that is passed from one time step to the next.

Each hidden state contains information from previous inputs

It is updated recursively at every time step

This allows the network to retain and use information from earlier parts of the sequence.

✅ 3. What challenges arise when backpropagating through long sequences?
Answer:

The main challenges are:

1. Vanishing Gradient Problem
Gradients become very small during backpropagation

Model fails to learn long-term dependencies

2. Exploding Gradient Problem
Gradients become very large

Training becomes unstable

👉 These issues occur because gradients are repeatedly multiplied across many time steps (Backpropagation Through Time).

✅ 4. What are the limitations of basic RNN-based language models?
Answer:

Basic RNNs have several limitations:

1. Difficulty capturing long-term dependencies
Memory fades over long sequences

2. Vanishing gradient problem
Makes training ineffective for deep sequences

3. Slow computation
Sequential processing → cannot fully parallelize

4. Limited context understanding
Cannot capture complex relationships effectively

👉 Because of these limitations, advanced architectures are used:

LSTM (Long Short-Term Memory)

GRU (Gated Recurrent Unit)
